```python
user_input=input("Enter the input of string==")

is_iter= True
while is_iter :
    print("\n===== STRING ANALYZER =====")
    print("1. Count characters")
    print("2. Count words")
    print("3. Uppercase")
    print("4. Lowercase")
    print("5. Swap case")
    print("6. Search for a word")
    print("7. Replace a word")
    print("8. Reverse text")
    print("9. Check palindrome")
    print("10. Exit")
    choice = int(input("\nEnter your choice as option from 1-10:" ))
    print("\n")

    if choice >=1 and choice <=10:
        if choice == 1:
            print("No of characters in sentence : ",len(user_input))
        elif choice ==2:
            print("No of words present in sentence :",len(user_input.split()))
        elif choice ==3:
            print("Upper case of entire sentence :",user_input.upper())
        elif choice ==4:
            print("Lower case of entire sentence :",user_input.lower())
        elif choice ==5:
            print("Swapcase of entire sentence :",user_input.swapcase())
        elif choice ==6:
            search_word=input("Enter the word which you want to search :")
            if search_word.lower() in user_input.lower().split():
                print("The {} will be located on index of :".format(search_word,user_input.find(search_word)))
            else:
                print("{} is not found in given user input of {}".format(search_word,user_input))
        elif choice ==7:
            replace_word=input("Enter the word which you want to replace :")
            new_word=input("Enter the word which you want to add :")
            print("The updated sentence :",user_input.replace(replace_word,new_word))
        elif choice ==8:
            print("The reversed text of given sentence :",user_input.lower()[::-1])
        elif choice ==9:
            for word in user_input:
                if word == user_input[::-1]:
                    print("{} is palandrome".format(user_input))
        else:
            print("Exiting String Analyzer...")
            is_iter = False
    else:
        print("Please enter the correct option between 1-10")
```
